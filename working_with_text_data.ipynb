```python
#Simple Implementation of a simple tokenizer class.Version 1
class SimpleTokenizerV1:

    def __init__(self, vocab):
        # pass
        self.str_to_int = vocab
        self.int_to_str = {i:s for s,i in vocab.items()}

    def encode(self, text):

        preprocessed = re.split(r'([.,;:?!()_%/\s"\']|--)', text)
        preprocessed = [item for item in preprocessed if item.strip()]
        ids = [self.str_to_int[token] for token in preprocessed]
        return ids
    def decode(self, ids:List):

        #join words/tokens with a s[ace]
        text = " ".join([self.int_to_str[token_id] for token_id in ids])

        #Here we are 
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)

        return text
```

```python
#Tokenizing the text.
tokenizer = SimpleTokenizerV1(vocab_dict)
text = """It's the last he painted, you know," Mrs. Gisburn said"""
ids = tokenizer.encode(text)
print(ids)
```

```python
#Lets see if we can decode the text.
recovered_text = tokenizer.decode(ids=ids)
print(recovered_text)
```

```python
#Simple Tokenizer Version 2
class SimpleTokenizerV2:

    def __init__(self, vocab):
        # pass
        self.str_to_int = vocab
        self.int_to_str = {i:s for s,i in vocab.items()}

    def encode(self, text):

        preprocessed = re.split(r'([.,;:?!()_%/\s"\']|--)', text)
        preprocessed = [item for item in preprocessed if item.strip()]
        #modified from version 1 . 
        #Here we are replacing unknown words with id pf the <|unk|> token.
        ids = [self.str_to_int[token] if token in self.str_to_int.keys() else self.str_to_int["<|unk|>"] for token in preprocessed ]
        #The following also works.
        # preprocessed = [item if item in self.str_to_int else "<|unk|>" for item in preprocessed]
        # ids = [self.str_to_int[s] for s in preprocessed]
        return ids
    def decode(self, ids:List):

        #join words/tokens with a s[ace]
        text = " ".join([self.int_to_str[token_id] for token_id in ids])

        #Here we are substituing the space followed by punctuation or special characters with only punctuation because Hi! Okay is right not Hi ! Okay.
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)

        return text
```

```python
import torch
from torch.utils.data import Dataset, DataLoader
```

```python
class GPTDataset_V1(Dataset):
    #For a class that inherits from abstract class o Dataset we need to incorporate __init_(), __len__() & __getitem__() methods.
    def __init__(self, text_corpus, tokenizer, max_length, stride=1):
        self.tokenizer = tokenizer
        self.input_ids = []
        self.target_ids = []

        token_ids = self.tokenizer.encode(text_corpus)

        #Creating the dataset.
        for i in range(0, len(token_ids) - max_length, stride):  #Here we will by default use 1 stride for our text corpus.
            input_chunk = token_ids[i:i+max_length]
            target_chunk = token_ids[i+1:i+max_length+1]
            self.input_ids.append(torch.tensor(input_chunk))
            self.target_ids.append(torch.tensor(target_chunk))

    def __len__(self):
        return  len(self.input_ids)

    def __getitem__(self, idx):
        return  self.input_ids[idx], self.target_ids[idx]
```

```python
#Checking the dataset only.
text_dataset = GPTDataset_V1(text_corpus=raw_text, tokenizer=tokenizer_gpt, max_length=4, stride=1)
#Fetching a particular input-target pair.
text_dataset.__getitem__(idx=5) #We can also do text_dataset[5] same thing.
```

```python
#> For multiple indices fetching in Dataset.
text_dataset[1:3]
```

```python
#Now we implement the DataLoader class.
def create_dataloader_v1(text, max_length, stride, batch_size, shuffle):

    gpt_tokenizer = tiktoken.get_encoding('gpt2')
    dataset = GPTDataset_V1(text_corpus=text, tokenizer=gpt_tokenizer, max_length=max_length, stride=stride)
    dataloader = DataLoader(dataset=dataset, batch_size=batch_size, shuffle=shuffle, drop_last=True)

    return dataloader
```

```python
with open('the-verdict.txt', 'r') as f:
    raw_text = f.read()
#Using the DataLoader .    
text_dataloader_stride1 = create_dataloader_v1(text=raw_text, max_length=4, stride=1, batch_size=4, shuffle=False) #Shuffling shuffles the batches.
data_iter = iter(text_dataloader_stride1)
first_batch_input, first_batch_target = next(data_iter)
print(f'Stride 1 result\nFirst Batch Input (x):\n\n{first_batch_input}\n\nFirst Batch Target(y):\n{first_batch_target} & \nBatch Size:{len(first_batch_input)}\nBatch Shape:{first_batch_input.shape}')
```

### 2.8 Encoding word positions

> There are 2 types of Postional Embeddings: 1. Absolute Positional Embeddings 2. Relative Positional Embeddings.

> In Early Gpt Models OpenAI used Absoulte Positional Embeddings but they were also trained and updated and optimized during training mechanism.

> Absolute Positional Embeddings Image.

![Alt Text](pics/Absolute_Positional_Embedding.png)

```python
#Implemeting Positional Embeddings.
#Let's Imagine we have Vocab size of 50257 and output_dim of 256.
output_dim, vocab_size = 256, 50257
#So our token embedding for our entire vocabulary is:
token_embedding_layer = torch.nn.Embedding(num_embeddings=vocab_size, embedding_dim=output_dim)

#Now we need to create Absolute Positional Embeddings for our tokensbut with same output dim.
max_length = 4
dataloader = create_dataloader_v1(text=raw_text, max_length=max_length, stride=max_length, batch_size=8, shuffle=True)

data_iter = iter(dataloader)
batch_inputs, batch_targets = next(data_iter)
print("Token IDs:\n", batch_inputs)
print("\nInputs shape:\n", batch_inputs.shape)
```

> Now the data is 8 x 4 size that is a batch of 8 inputs with context window of 4.

> Now we need to get the initialized embeddings for this 1 batch oonly which should be 8 x 4 x 256 as the embedding/output dim is 256.

```python
#Token Embedding.
token_embeddings = token_embedding_layer(batch_inputs)  #Getting the Embedding of the batch input tokens of size 8 x 4
print("One batch embedding shape:",token_embeddings.shape)
```

> Absolute Positional Embedding Creation

```python
#Positional EMbedding.
context_length = max_length
postional_embedding_layer = torch.nn.Embedding(num_embeddings=context_length, embedding_dim=output_dim)

#This code : torch.arange(4) produces: tensor([0, 1, 2, 3]) which we need to get postional embedding vector intialization
positional_embeddings = postional_embedding_layer(torch.arange(context_length))
print('Positional Embedding Shape:',positional_embeddings.shape)
```

```python
#Input Embedding.
#Now we need to do token embedding vector + postional embedding vector -> Input Embedding.
input_embeddings= token_embeddings + positional_embeddings
print('Input Embedding shape',input_embeddings.shape)
```
